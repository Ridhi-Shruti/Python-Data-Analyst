```python
zo['rating_category'] = pd.cut(
    zo['rate'],
    bins=[0, 3, 4, 5],
    labels=['Low', 'Average', 'High'],
    right=False
)
```

```python
zo['rating_category'].value_counts()
```

```python
zo['cost_category'] = pd.cut(
    zo['approx_cost(for two people)'],
    bins=[0, 300, 600, float('inf')],
    labels=['Budget', 'Moderate', 'Expensive'],
    right=False
)
```

```python
zo['cost_category'].value_counts()
```

```python
zo['engagement_category'] = pd.cut(
    zo['votes'],
    bins=[-1, 100, 500, float('inf')],
    labels=['Low', 'Medium', 'High']
)
```

```python
zo['engagement_category'].value_counts()
```

```python

plt.figure(facecolor='lightgreen',figsize=(8,5))

count=sns.countplot(
    data=zo,
    x='listed_in(type)',
    order=zo['listed_in(type)'].value_counts().index,
    hue='rating_category',
    palette='twilight',
    width=0.5
)

plt.title('Distribution of Restaurant Types')
plt.xlabel('Restaurant Type')
plt.ylabel('Number of Restaurants')
plt.gca().set_facecolor('lightblue')


for i in count.containers:
  count.bar_label(i)
plt.show()
```

# Average Rating by Restaurant Type

```python
rating_by_type = zo.groupby('listed_in(type)')['rate'].mean().sort_values(ascending=False)

plt.figure(facecolor='seagreen',figsize=(8,5))

bar=sns.barplot(
    x=rating_by_type.index,
    y=rating_by_type.values,
    hue=rating_by_type.index,
    palette='twilight',
    edgecolor='black',
    linewidth=1,
    width=0.5
)

for i in bar.containers:
  bar.bar_label(i)


plt.title('Average Rating by Restaurant Type')
plt.xlabel('Restaurant Type')
plt.ylabel('Average Rating')
plt.ylim(0, 5)

plt.gca().set_facecolor('powderblue')

plt.show()
```

# Online Order vs Average Rating

```python
order_rating = zo.groupby('online_order')['rate'].mean()

plt.figure(facecolor='khaki', figsize=(7,5))

bar=sns.barplot(
    x=order_rating.index,
    y=order_rating.values,
    hue=order_rating.index,
    palette='muted',
    edgecolor='black',
    linewidth=1,
    width=0.4
)

plt.title('Average Rating by Online Order Availability')
plt.xlabel('Online Order')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.gca().set_facecolor('teal')
for i in bar.containers:
  bar.bar_label(i)

plt.show()
```

# Table Booking vs Average Rating

```python
booking_rating = zo.groupby('book_table')['rate'].mean()

plt.figure(facecolor='skyblue',figsize=(7,5))

bar=sns.barplot(
    x=booking_rating.index,
    y=booking_rating.values,
    hue=booking_rating.index,
    palette='colorblind',
    edgecolor='black',
    linewidth=1,
    width=0.4
)

for i in bar.containers:
  bar.bar_label(i)

plt.title('Average Rating by Table Booking Availability')
plt.xlabel('Table Booking')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.gca().set_facecolor('moccasin')

plt.show()
```

# Cost Category vs Average Rating 💰

```python
cost_rating = zo.groupby('cost_category', observed=True)['rate'].mean().sort_values(ascending=False)

plt.figure(facecolor='peachpuff',figsize=(8,5))

bar=sns.barplot(
    x=cost_rating.index,
    y=cost_rating.values,
    hue=cost_rating.index,
    palette='pastel',
    edgecolor='black',
    linewidth=1,
    width=0.4
)

for i in bar.containers:
  bar.bar_label(i)


plt.title('Average Rating by Cost Category')
plt.xlabel('Cost Category')
plt.ylabel('Average Rating')
plt.ylim(0, 5)
plt.gca().set_facecolor('khaki')
plt.show()
```

# Votes vs Rating

```python
plt.figure(facecolor='lightcyan',figsize=(8,5))

sns.scatterplot(
    data=zo,
    x='votes',
    y='rate',
    alpha=0.6,
    color='blue'
)

plt.title('Relationship Between Votes and Ratings')
plt.xlabel('Number of Votes')
plt.ylabel('Rating')
plt.gca().set_facecolor('lightyellow')

plt.show()
```
